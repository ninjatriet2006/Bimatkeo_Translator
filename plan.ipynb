#### 📋 Chi tiết triển khai & Giao tiếp:
- **Cấu hình & Tùy biến (Configuration)**: UI quản lý cấu hình người dùng thông qua các tệp cấu hình YAML đặt tại `.config/configs/`. - **Cách Gọi (Call)**: Giao diện GUI sinh một tệp cấu hình JSON tạm thời trong thư mục `temp/task_[timestamp].json` chứa danh sách ảnh cần dịch cùng các tham số mô hình đã được xác thực. Sau đó, GUI khởi chạy CLI bằng subprocess với tham số truyền vào đường dẫn tệp JSON này:
```python
# Minh họa cơ chế GUI gọi CLI Backend độc lập
import subprocess
import os
import json
import time

task_data = {
    "profile": "nllb_manga",
    "src_lang": "ja",
    "tgt_lang": "vi",
    "batch_size": 4,
    "images": ["/absolute/path/to/page1.png", "/absolute/path/to/page2.png"]
}

task_file = f"temp/task_{int(time.time())}.json"
os.makedirs("temp", exist_ok=True)
with open(task_file, "w", encoding="utf-8") as f:
    json.dump(task_data, f, ensure_ascii=False, indent=4)

# Khởi chạy CLI chạy nền, cô lập nhóm tiến trình con
process = subprocess.Popen(
    ["python", "app/core/pipeline.py", "--task", task_file],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True,
    preexec_fn=os.setsid  # Chỉ chạy trên Linux: Tách nhóm tiến trình để tránh dính zombie process
)
```
- **Cách Nhận kết quả (Receive)**: Tiến trình UI không chặn (non-blocking) các sự kiện bằng cách khởi chạy một luồng Daemon Thread để liên tục đọc đầu ra chuẩn stdout từ tiến trình con. Khi phát hiện các nhãn tiến độ định dạng `[PROG]`, luồng này sẽ cập nhật giao diện thông qua cơ chế Signal/Slot của Qt:
```python
# Luồng nền giám sát stdout và cập nhật tiến độ lên UI
import threading

def monitor_stdout(process, signal_emitter):
    for line in iter(process.stdout.readline, ""):
        line = line.strip()
        if line.startswith("[PROG]"):
            # Ví dụ: "[PROG][OCR] [##--------] 20% - Đang quét chữ trang 1"
            parts = line.split("]", 2)
            stage = parts[1].replace("[", "")
            progress_desc = parts[2].strip()
            # Phát tín hiệu an toàn về Main GUI Thread
            signal_emitter.emit(stage, progress_desc)
    process.wait()
```

#### 📋 Chi tiết triển khai & Giao tiếp:

```python
import argparse
import json

parser = argparse.ArgumentParser()
```

#### 📋 Đặc tả Lớp giao tiếp trừu tượng (Abstract Base Classes - ABC):
```python
from abc import ABC, abstractmethod
import numpy as np

class BaseOCR(ABC):
    @abstractmethod
    def load_model(self, model_path: str) -> None:
        """Tải mô hình phát hiện & nhận diện chữ (như Manga-OCR, PaddleOCR) lên thiết bị suy luận."""
        pass

    @abstractmethod
    def detect_and_recognize(self, image: np.ndarray) -> tuple[list[list[int]], list[str]]:
        """Quét chữ trên mảng ảnh numpy đầu vào, trả về: (danh sách bboxes [x_min, y_min, x_max, y_max], danh sách text gốc)."""
        pass

class BaseTranslator(ABC):
    @abstractmethod
    def load_weights(self, model_path: str) -> None:
        """Tải trọng số mô hình dịch thuật (NLLB, M2M100...) từ đường dẫn được chỉ định."""
        pass

    @abstractmethod
    def translate(self, texts: list[str], src_lang: str, tgt_lang: str) -> list[str]:
        """Dịch danh sách các đoạn text tiếng nguồn sang tiếng đích."""
        pass

class BaseInpainter(ABC):
    @abstractmethod
    def load_model(self, model_path: str) -> None:
        """Tải mô hình inpainting (Lama, PatchMatch...) lên bộ nhớ."""
        pass

    @abstractmethod
    def inpaint(self, image: np.ndarray, bboxes: list[list[int]]) -> np.ndarray:
        """Nhận ảnh gốc và tọa độ bboxes chứa văn bản, thực hiện xóa chữ và trả về mảng ảnh nền sạch."""
        pass

class BaseRenderer(ABC):
    @abstractmethod
    def load_fonts(self, font_path: str) -> None:
        """Cấu hình và nạp font chữ TrueType/OpenType hỗ trợ viết chữ."""
        pass

    @abstractmethod
    def render(self, image: np.ndarray, bboxes: list[list[int]], texts: list[str]) -> np.ndarray:
        """Vẽ văn bản tiếng đích (texts) lên ảnh nền đã làm sạch (image) tại đúng vị trí bboxes."""
        pass
```

#### 📋 Cơ chế nạp động Factory Registry:
```python
# Triển khai Factory tự động đăng ký dynamic subclass thông qua decorator
class TranslatorFactory:
    _registry = {}

    @classmethod
    def register(cls, name: str):
        """Decorator dùng để đăng ký một triển khai dịch thuật cụ thể."""
        def decorator(subclass):
            cls._registry[name] = subclass
            return subclass
        return decorator

    @classmethod
    def create(cls, name: str, model_path: str) -> BaseTranslator:
        """Tạo và trả về instance của lớp triển khai dịch thuật."""
        if name not in cls._registry:
            raise ValueError(f"Mô hình dịch '{name}' chưa được đăng ký vào Factory.")
        impl_class = cls._registry[name]
        instance = impl_class()
        instance.load_weights(model_path)
        return instance

# Ví dụ đăng ký NLLB-200
@TranslatorFactory.register("nllb-200")
class NLLBTranslator(BaseTranslator):
    def load_weights(self, model_path: str):
        # Code tải weights dùng HuggingFace hoặc CTranslate2
        pass

    def translate(self, texts: list[str], src_lang: str, tgt_lang: str) -> list[str]:
        # Logic thực thi dịch thuật offline
        return [f"[Dịch NLLB] {t}" for t in texts]
```

#### 🧬 Định nghĩa cấu trúc PageContext DTO trên RAM
```python
from dataclasses import dataclass
import numpy as np

@dataclass
class PageContext:
    page_id: str                 # Định danh trang (Ví dụ: Chap_1.1_001)
    original_image: np.ndarray   # Mảng ảnh gốc dạng NumPy lưu trên RAM
    bboxes: list = None          # Tọa độ khung chữ (từ Worker OCR)
    original_texts: list = None  # Văn bản gốc nhận diện được (từ Worker OCR)
    translated_texts: list = None# Văn bản đã dịch (từ Worker Translator)
    text_styles: list = None     # Định dạng chữ (Màu sắc, kích cỡ, font từ OCR phân tích)
    inpainted_image: np.ndarray = None # Mảng ảnh đã xóa nền bong bóng thoại
```

#### 📋 Triển khai luồng Worker chạy song song gối đầu (Pipeline Threads):
Mỗi luồng Worker chịu trách nhiệm gọi một dịch vụ AI, thực hiện suy luận, cập nhật trường dữ liệu tương ứng trong DTO `PageContext` trên RAM và đẩy sang Queue tiếp theo mà không lưu bất kỳ file ảnh trung gian nào:
```python
import threading
import queue

class OCRWorker(threading.Thread):
    def __init__(self, in_q: queue.Queue, out_q: queue.Queue, ocr_engine: BaseOCR):
        super().__init__()
        self.in_q = in_q
        self.out_q = out_q
        self.ocr_engine = ocr_engine
        self.daemon = True

    def run(self):
        while True:
            # Lấy PageContext từ RAM Queue
            ctx: PageContext = self.in_q.get()
            if ctx is None:  # Nhận tín hiệu dừng (Sentinel)
                self.out_q.put(None)
                break
            
            # Gọi trực tiếp qua interface trừu tượng
            bboxes, texts = self.ocr_engine.detect_and_recognize(ctx.original_image)
            ctx.bboxes = bboxes
            ctx.original_texts = texts
            
            # Đẩy kết quả trung gian sang hàng đợi tiếp theo (Zero-Copy RAM)
            self.out_q.put(ctx)
            self.in_q.task_done()
```

### 5️⃣ Quản lý Tài nguyên (Logic Chia lô & Giải phóng VRAM)
Để tránh lỗi tràn bộ nhớ đồ họa (Out of Memory - OOM), thuật toán tự động đếm tổng số ảnh thực tế ($Total$) và tính toán chia nhỏ thành $n$ lô sao cho lượng ảnh nạp vào GPU không vượt quá giới hạn:

$$\text{Total} / n \le \text{Max\_Batch\_Size}$$

Sau khi hoàn tất mỗi lô ảnh, `Worker Inpainting` hoặc luồng thu hồi tài nguyên chủ động dọn dẹp bộ nhớ đồ họa bằng lệnh:
```python
# Triển khai giải phóng bộ nhớ sau mỗi lô xử lý nhằm ngăn tràn VRAM (OOM)
import torch  # type: ignore
import gc

def release_gpu_memory():
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    gc.collect()
```
